```python
#palidram-->reads the same forward and backword
string = input("give a string: ")
reverse=string[::-1]
if reverse==string:
  print("this is palidram")
else:
  print("this is not palidram")
```

```python
num = input("Give a num value: ")

reverse = num[::-1]

if reverse == num:
    print("This is a palindrome")
else:
    print("This is not a palindrome")
```

```python
String = input("Give string value: ")

reverse = string[::-1]

if reverse == string:
    print("This is a palindrome")
else:
    print("This is not a palindrome")
```
